```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image in grayscale format
img = cv2.imread('cameraman.tif', cv2.IMREAD_GRAYSCALE)

# Convert the image to float32 and normalize pixel values to [0, 1]
img_float = img.astype(np.float32) / 255.0
```

```python
# Apply a 7x7 Box filter for smoothing
box_filtered = cv2.boxFilter(img, -1, (7, 7))

# Display the original and smoothed images side by side
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

# Plot the original image
ax[0].imshow(img, cmap='gray')
ax[0].set_title("Original Image")
ax[0].axis("off")

# Plot the smoothed image
ax[1].imshow(box_filtered, cmap='gray')
ax[1].set_title("Smoothed (7x7 Box Filter)")
ax[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Apply Gaussian filter with 5x5 kernel size
gauss_5x5 = cv2.GaussianBlur(img, (5, 5), sigmaX=0)

# Apply Gaussian filter with 21x21 kernel size
gauss_21x21 = cv2.GaussianBlur(img, (21, 21), sigmaX=0)

# Display the three images side by side
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Plot original image
ax[0].imshow(img, cmap='gray')
ax[0].set_title("Original Image")
ax[0].axis("off")

# Plot 5x5 Gaussian blur
ax[1].imshow(gauss_5x5, cmap='gray')
ax[1].set_title("Gaussian Blur (5x5)")
ax[1].axis("off")

# Plot 21x21 Gaussian blur
ax[2].imshow(gauss_21x21, cmap='gray')
ax[2].set_title("Gaussian Blur (21x21)")
ax[2].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Apply 3x3 Laplacian filter to detect edges
laplacian = cv2.Laplacian(img_float, cv2.CV_32F, ksize=3)

# Sharpen the image by subtracting the Laplacian and clipping values between 0 and 1
sharpened = np.clip(img_float - laplacian, 0, 1)

# Display original, Laplacian, and sharpened images side by side
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Plot original image
ax[0].imshow(img_float, cmap='gray')
ax[0].set_title("Original Image")
ax[0].axis("off")

# Plot Laplacian edge map
ax[1].imshow(laplacian, cmap='gray')
ax[1].set_title("Laplacian (Edges)")
ax[1].axis("off")

# Plot sharpened image
ax[2].imshow(sharpened, cmap='gray')
ax[2].set_title("Sharpened Image")
ax[2].axis("off")

plt.tight_layout()
plt.show()
```
