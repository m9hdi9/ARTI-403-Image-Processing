```python
import cv2
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np
```

```python
# قراءة الصورة
img = cv2.imread('images/cameraman.tif')

# عرض الصورة
plt.imshow(img)
plt.title("Loaded with OpenCV")
plt.axis('off')
plt.show()
```

```python
# فتح الصورة
img2 = Image.open('images/lena_gray_256.tif')

# عرض الصورة بتدرج رمادي
plt.imshow(img2, cmap=cm.Greys_r)
plt.title("Loaded with PIL")
plt.axis('off')
plt.show()
```

```python
# الحفظ باستخدام OpenCV
cv2.imwrite('new_image.jpg', img)

# الحفظ باستخدام PIL
img2.save('new_image2.jpg')
```

```python
# صورة OpenCV تكون مصفوفة NumPy بشكل تلقائي
print("OpenCV Image Shape:", img.shape)
print("OpenCV Image Data Type:", img.dtype)
print("OpenCV Image Array:\n", img)

print("-" * 50)

# صورة PIL يلزم تحويلها أولاً إلى NumPy array
img_array = np.array(img2)
print("PIL Image Shape:", img_array.shape)
print("PIL Image Data Type:", img_array.dtype)
print("PIL Image Array:\n", img_array)
```

```python
# معرفة أقل وأعلى قيمة للبكسل
print("Min pixel value:", img_array.min())
print("Max pixel value:", img_array.max())

# حساب متوسط قيم البكسلات
print("Mean pixel value:", img_array.mean())

# أخذ جزء مقتطع من الصورة (Slicing/Cropping)
crop = img_array[50:150, 50:150]
plt.imshow(crop, cmap='gray')
plt.title("Cropped Region")
plt.axis('off')
plt.show()
```
